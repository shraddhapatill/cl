```python
# Experiment: Document Retrieval using Inverted Files

import os

# --------------------------------------------------
# METHOD 1: Read documents from a folder
# --------------------------------------------------

# Folder containing text documents
# folder = "IR2"

# Dictionary to store document contents
# documents = {}

# Read all .txt files
# for filename in os.listdir(folder):
#     if filename.endswith(".txt"):
#         filepath = os.path.join(folder, filename)
#
#         with open(filepath, "r", encoding="utf-8") as file:
#             documents[filename] = file.read().lower()


# --------------------------------------------------
# METHOD 2: Create documents using a dictionary
# --------------------------------------------------

documents = {
    "doc1.txt": "Information retrieval is used to search documents.",
    "doc2.txt": "Search engines help retrieve relevant information.",
    "doc3.txt": "Python is used for information retrieval."
}

# --------------------------------------------------
# CREATE INVERTED INDEX
# --------------------------------------------------

inverted_index = {}

for filename, text in documents.items():
    words = text.lower().split()

    for word in words:
        # Remove punctuation
        word = word.strip(".,!?;:")

        if word not in inverted_index:
            inverted_index[word] = []

        if filename not in inverted_index[word]:
            inverted_index[word].append(filename)

# --------------------------------------------------
# DISPLAY INVERTED INDEX
# --------------------------------------------------

print("INVERTED INDEX")

for word in sorted(inverted_index):
    print(word, "->", inverted_index[word])

# --------------------------------------------------
# SEARCH DOCUMENTS
# --------------------------------------------------

query = input("\nEnter search query: ").lower().strip()

query_words = query.split()

# Find documents containing all query words
if query_words:
    result = set(inverted_index.get(query_words[0], []))

    for word in query_words[1:]:
        result = result.intersection(
            inverted_index.get(word, [])
        )

    if result:
        print("\nDocuments found:")

        for filename in sorted(result):
            print(filename)
    else:
        print("\nNo documents found.")
else:
    print("\nPlease enter a search query.")
```
