```python
# ASS 4: Genomic Machine Learning

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Step 1: Load dataset
df = pd.read_csv("D:/Sem_7/BIF Lab/Genomic ML/data/genomic_data.csv")

# Step 2: Display column names
print("Columns:", df.columns.tolist())

# Step 3: Select features and target
X = df[["Gene One", "Gene Two"]]
y = df["Cancer Present"]

# Step 4: Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Step 5: Create and train Random Forest model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Step 6: Predict cancer presence
y_pred = model.predict(X_test)

# Step 7: Evaluate model performance
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Step 8: Display feature importance
print("\nFeature Importance:")

for col, imp in zip(X.columns, model.feature_importances_):
    print(col, ":", round(imp, 3))

# Step 9: Display analysis report
print("\nAnalysis Report:")
print("Algorithm: Random Forest Classifier")
print("Features: Gene One and Gene Two")
print("Target: Cancer Present")
print("Training data:", len(X_train), "records")
print("Testing data:", len(X_test), "records")
print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nConclusion:")
print("The Random Forest model classifies genomic data")
print("based on gene features and predicts cancer presence.")
```
